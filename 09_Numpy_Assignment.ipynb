````python
import numpy as np

# Store names
stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

# Product categories
categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

# Sales revenue in rupees
# Rows represent stores
# Columns represent product categories

sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

# Cost incurred in rupees
# Rows represent stores
# Columns represent product categories

cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])

print("Stores:")
print(stores)

print("\nProduct Categories:")
print(categories)

print("\nSales Data:")
print(sales_data)

print("\nCost Data:")
print(cost_data)
````

## Business Observations

1. **Mumbai is the strongest overall performer** — it leads in total sales, total profit, and units sold, and comfortably clears the expansion thresholds on sales, profit and margin.
2. **Pune consistently lags the other stores**, showing the lowest total profit and the lowest number of category targets achieved, which points to store-level (not category-level) issues such as footfall or local pricing.
3. **Electronics drives the most revenue but not the best margin** — it is the top category by sales and profit in absolute rupees, but its profit margin trails smaller categories like Groceries and Beauty, so high volume is not translating into proportionally high profitability.
4. **Several store-category combinations fall below the 22% margin bar**, concentrated mainly in Electronics and Furniture — these are the same categories flagged for cost optimization, confirming that cost (not pricing) is the main margin drag.
5. **Inventory reorder needs are spread thin rather than concentrated in one store**, suggesting a general restocking cadence issue across the chain rather than a single store's supply-chain problem.

## Management Recommendations

1. **Prioritize Mumbai for expansion investment** — it is the only store that clears all three expansion criteria (sales, profit, and margin), making it the lowest-risk place to add footprint or headcount first.
2. **Renegotiate supplier costs for Electronics and Furniture** before scaling volume further — both categories repeatedly show high cost against sub-25% margins, so growing sales there without fixing cost will just grow low-margin revenue.
3. **Increase promotional spend on Beauty and other high-margin, low-average-sales categories** — they already convert well on margin, so extra marketing rupees are more likely to translate directly into profit than the same spend on Electronics.
